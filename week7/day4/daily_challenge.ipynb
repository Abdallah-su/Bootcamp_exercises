```python
from scipy.ndimage import rotate
import numpy as np
def rotate_image_30_degrees(image):
    img_array = np.array(image)
    rotated_img_array = rotate(img_array, 30, reshape=False)
    return Image.fromarray(rotated_img_array) #revert back to image format

rotated_img = rotate_image_30_degrees(img)
plt.imshow(rotated_img)
plt.axis('off')  # Hide axes
plt.show()
```

```python
from PIL import ImageOps
def flip_image_horizontally(image):
    return ImageOps.mirror(image)
def flip_image_vertically(image):
    return ImageOps.flip(image)
horizontally_flipped_img = flip_image_horizontally(img)
vertically_flipped_img = flip_image_vertically(horizontally_flipped_img)
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
cols=['horizontally flipped', 'vertically flipped']
for ax, img, col in zip(axes, [horizontally_flipped_img, vertically_flipped_img], cols):
    ax.imshow(img)
    ax.set_title(col)
    ax.axis('off')  # Hide axes
plt.tight_layout()
plt.show()
```

```python
def zoom_image(image, scale=1.2):
    width, height = image.size
    new_width, new_height = int(width * scale), int(height * scale)
    zoomed_image = image.resize((new_width, new_height), Image.LANCZOS)
    left = (new_width - width) // 2
    top = (new_height - height) // 2
    right = left + width
    bottom = top + height
    return zoomed_image.crop((left, top, right, bottom))
zoomed_img = zoom_image(img)
plt.imshow(zoomed_img)
plt.axis('off')  # Hide axes        
plt.show()
```
