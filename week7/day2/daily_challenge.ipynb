The Story: Using Data to Combat the 2014-2016 Ebola Outbreak.
• The Data:
 Mobile phone location data (anonymized) and transmission heatmaps.
• The Role of Analysis:
Data scientists analyzed movement patterns to predict where the virus would strike next.
In the 2014 Ebola case, the analysis was a blend of traditional epidemiology and modern big data analytics. Because the virus was moving so fast, static spreadsheets weren't enough.
Here are the specific types of analysis performed:
1. Spatiotemporal Analysis (The "Where" and "When")
Analysts used geographic information systems (GIS) and GPS data to map the outbreak in real-time.
• Method: They combined gravity-type models (which predict how people move between large population centers) with local transmission data.
• What was analyzed: They looked at "weekly incidence rates" (new cases per week) across different districts.
• Result: This helped identify "hotspots" before they became unmanageable, allowing for targeted quarantine zones and resource drops.
2. Social Network Analysis (The "Who")
To understand how the virus jumped from person to person, analysts used Network Science.
• Method: Using call detail records (CDRs) from mobile phones, they built "connectivity maps."
• What was analyzed: They didn't listen to calls, but they looked at the metadata: who was calling whom and which cell towers their phones were pinging.
• Result: This created a digital version of contact tracing. If a person tested positive, analysts could see which villages they had visited and which communities were most "connected" to them, flagging those areas for high risk.
3. Predictive Mathematical Modeling (The "What Next")
This involved using the SEIR Model, a classic data science framework for infectious diseases.
• The Framework: * Susceptible (People who could get it)
• Exposed (People who have it but aren't yet infectious)
• Infectious (People currently spreading it)
• Recovered/Removed (People no longer in the pool)
• Analysis: They calculated the \bm{R_0} (Basic Reproduction Number) to determine if the outbreak was growing or shrinking.
4. Sentiment and Behavioral Analysis
Surprisingly, Social Media Data (Twitter and Facebook) was analyzed using Natural Language Processing (NLP).
• The Goal: To measure public awareness and misinformation.
• Result: By analyzing the sentiment and "misinformation clusters" in specific regions, health organizations knew where they needed to send educational teams to combat rumors about the disease. This allowed organizations like the CDC and WHO to deploy medical supplies and personnel to specific villages before the outbreak peaked there.
• The Impact:
It shifted the strategy from reactive to predictive, likely saving thousands of lives by containing the spread in high-risk corridors.
