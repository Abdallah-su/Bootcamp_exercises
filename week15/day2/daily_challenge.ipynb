```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from statsmodels.stats.power import NormalIndPower
```

Calculate the Required Sample Size:
Imagine the current checkout process has a conversion rate of 5%, and the bakery’s team is confident the new process will boost this to 7%.
Use Python to calculate the required sample size per group (current vs. new checkout process) with an effect size of 0.2, a significance level of 0.05, and a desired power of 0.8.



```python
power = NormalIndPower()
s_sample = power.solve_power(effect_size=0.2, alpha=0.05, power=0.8, alternative='two-sided')
print(f"Required sample size for a two-sided test with effect size 0.2, alpha 0.05, and power 0.8: {s_sample:.0f}")
```

Analyze the Impact of Effect Size:

The bakery’s head chef, always aiming for perfection, wonders what would happen if the effect size were different. Calculate the required sample size for effect sizes of 0.1, 0.2, 0.3, and 0.4.
Explain how the sample size requirements change as the effect size changes. Why does this happen?



```python
effect_list =[0.1, 0.2, 0.3, 0.4]
sample_list=[]
for effect in effect_list:
    sample_size = power.solve_power(effect_size=effect, alpha=0.05, power=0.8, alternative='two-sided')
    sample_list.append(sample_size)
    print(f"Effect Size: {effect}, Required Sample Size: {sample_size:.0f}")
```

The lower the effect-size, the larger the sample size required in order to have at least a test power of 80%.

Explain the Relationship:

Imagine you’re explaining this to the bakery’s team in a fun, easy-to-understand way. Why is it so important to balance effect size and sample size when planning an A/B test? Help them understand how this ensures they’re not wasting time or resources and how it helps them confidently make decisions that could increase their sweet sales.

Think of an A/B test like a magnifying glass and the "effect" (the change you’re looking for) like a tiny grain of sand on a beach.
1. The Magnifying Glass (Sample Size)
The more people you include in your test (Sample Size), the stronger your magnifying glass becomes.
• If you have a huge sample, you can see even the tiniest speck of dust.
• If you have a tiny sample, you can only see big objects, like a bright red seashell.
2. The Grain of Sand (Effect Size)
The Effect Size is simply how big the change is.
• If you change a button color and it makes a massive difference in sales, that’s a "seashell." It's easy to spot.
• If you change the font slightly and it makes a minuscule difference, that’s a "grain of sand." It's very hard to see.
Why the Balance Matters
If the glass is too weak (Sample size is too small):
You might be standing right over a gold nugget (a great business idea), but because your magnifying glass isn't strong enough, you miss it. You conclude "nothing is there" simply because you didn't look closely enough.
If the glass is too strong (Sample size is too big):
You will start seeing every microscopic speck of dust on the beach. You might get excited and say, "Look! I found something!" But in reality, that speck of dust is so small it doesn't actually matter to your business. You wasted a lot of time and money buying a super-powered microscope just to find dust.
The Goal: "Just Right"
You want a sample size that is just strong enough to see the things that actually matter to you.
• Big Changes: If you're looking for a big win, you can move fast with a small group of people.
• Small Tweaks: If you're at a massive scale where even a 0.1% change equals millions of dollars, you need a massive group of people and a very powerful magnifying glass.
