```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.barplot(data=dF.head(25), x='Country or region', y='Score')
plt.xticks(rotation=90)
plt.title('Happiness Score by Country (Top 25)') 
plt.show()
```

Afganistan, Centrak African Republic are the least happiest countries

```python
sns.boxplot(data=dF.head(25), x = 'Country or region', y = 'Score')
plt.xticks(rotation=90)
plt.title('Happiness Score by Country (Top 25)') 
plt.show()
```

```python
sns.boxplot(data =dF.tail(25), x = 'Country or region', y = 'Score')
plt.xticks(rotation=90)
plt.title('Happiness Score by Country (Bottom 25)') 
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.heatmap(dF.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.xticks(rotation= 90)
plt.show()
```

Part3:Interactive Visualizations with Plotly

Interactive World Map: Create an interactive map displaying happiness scores across different countries.
Dynamic Scatter Plot: Plot an interactive scatter plot exploring relationships between various factors like health, freedom, and happiness.
Factor Analysis: Use a bubble chart to visualize the impact of social support and generosity on happiness scores, with bubble size representing country population.


```python
import plotly.express as px
fig= px.choropleth(dF, locations='Country or region', 
                   locationmode='country names', color='Score', 
                   color_continuous_scale=px.colors.sequential.Plasma)
fig.update_layout(title='Happiness Score by Country')
fig.show()
```

```python
fig = px.scatter(dF, x='Healthy life expectancy', y='Score', 
                 hover_name='Country or region', color='Score',
                 color_continuous_scale=px.colors.sequential.Plasma,
                 animation_frame='Overall rank',
                 range_x=[dF['Healthy life expectancy'].min()-0.1,
                 dF['Healthy life expectancy'].max()+0.1], 
                 range_y=[dF['Score'].min()-0.1, dF['Score'].max()+0.1],
                text='Country or region', 
                hover_data={'Country or region':True, })
fig.update_layout(title='Happiness Score vs Healthy Life Expectancy') 
fig.show()   
```

```python


# 1. Load the inbuilt population data
gapminder = px.data.gapminder().query("year == 2007") # Use the latest year available
pop_data = gapminder[['country', 'pop']]

# 2. Merge it with your Happiness dataframe (df)
# This adds a 'pop' column to your data by matching country names
df = pd.merge(dF, pop_data, left_on='Country or region', right_on='country', how='left')
```

```python
gapminder.columns.tolist()
```

```python
df.columns.tolist()
```

```python

fig = px.scatter(df, x="Social support", y="Generosity", size="pop", color="country", 
                 hover_name="country", size_max=60, 
                 title="Generosity vs Social Support with Population Size")
fig.show()
```

```python
df['year']='2019'
df_15['year']= '2015'
df_15= df_15.rename(columns ={'Country':'country'})
df_15 =df_15.rename(columns= {'Happiness Score': 'Score'})
df_19_15= pd.concat([df, df_15], ignore_index = True)
```

```python
nations =[ 'Israel', 'Luxembourg', 'United Kingdom', 'Ireland', 'Germany', 'Belgium',
 'United States' ]
df_filted = df_19_15[df_19_15['country'].isin(nations)]
fig=px.line(df_filted, x =df_filted['year'], 
            y ='Score', color ='country', markers =True, 
            title ='Happiness score:2015, 2019 compared')
fig.show()
```

from the line plot, Germany,United Kingdom, Ireland had their happiness gone up whlist countries like Israel, United State saw a decline and Luxembourge saw a maginal decline.
