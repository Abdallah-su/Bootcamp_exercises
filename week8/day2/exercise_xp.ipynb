Exercise 1 : Understanding Data Visualization
Instructions
Explain why data visualization is important in data analysis.
Describe the purpose of a line graph in data visualization.




Data visualization is the bridge between raw numbers and human understanding. It is crucial because:  
• Identifying Patterns and Trends: It allows analysts to quickly spot upward or downward trends (like sales growth over time) that are difficult to see in a spreadsheet.  
• Spotting Outliers: Visuals make "weird" data points—like a single day with massive errors or unusually high sales—stand out immediately.  
• Simplifying Complex Data: It condenses thousands of rows of data into a single image, making it easier to share insights with people who aren't data experts.  
• Accelerating Decision Making: It helps stakeholders grasp the "big picture" instantly so they can act on the data faster. 


The Purpose of a Line Graph in Data Visualization
The primary purpose of a line graph is to show change over a continuous period of time.  
• Tracking Progression: It connects individual data points with a line to show the "flow" of data, making it the best tool for time-series analysis.
• Comparing Multiple Categories: You can plot multiple lines (like "Sales" vs. "Expenses") on the same graph to see how they interact over the same timeframe.  
• Visualizing Continuity: Unlike a bar chart, which focuses on individual categories, a line graph emphasizes the relationship and the rate of change between points.
