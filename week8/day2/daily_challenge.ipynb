```python
plt.scatter(x=df['Social support'], y = df['Score'])
plt.ylabel('Happiness Score')
plt.xlabel('Social Support')
plt.title("Correlation Between Social Support and Contries' Happiness Score")
plt.show()
```

```python
df['Country or region'].unique()
```

```python
fig, axes = plt.subplots(2, 1, figsize=(22,15))
axis_list= axes.flatten()
axis_list[0].bar(x =df['Country or region'], height =df['GDP per capita'])
axis_list[0].set_title('GDP Per Capita across regions')

axis_list[1].plot(df['Country or region'], df['Healthy life expectancy'])
axis_list[1].set_title('Healthy Life Trends Across Regions')
plt.tight_layout()
plt.show()
    
```

```python

fig, axes = plt.subplots(2, 1, figsize=(22,15))
axis_list= axes.flatten()
axis_list[0].bar(x =df[df['Overall rank'] <= 15]['Country or region'], height =df[df['Overall rank'] <= 15]['GDP per capita'])
axis_list[0].set_title('GDP Per Capita across regions')
axis_list[0].tick_params(axis ='both', which ='major', labelsize = 16, rotation= 45)
axis_list[1].plot(df[df['Overall rank'] <= 15]['Country or region'], df[df['Overall rank'] <= 15]['Healthy life expectancy'])
axis_list[1].set_title('Healthy Life Trends Across Regions')
axis_list[1].tick_params(axis ='both', which ='major', labelsize = 16, rotation =45)
plt.tight_layout()
plt.show()
    
```
