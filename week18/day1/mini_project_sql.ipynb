Query 1: Count and Percentage of Orders Purchased in Jan 2018 with 5 Review Score

# Write and execute a SQL query to count the number of orders purchased in January 2018 that have a review score of 5 and calculate the percentage of such orders.

```python
import pandas as pd

query ='''SELECT COUNT(oo.customer_id) AS order_count FROM olist_orders oo
WHERE oo.order_purchase_timestamp >= '2018-01-01' AND oo.order_purchase_timestamp < '2018-02-01' AND
(SELECT r.review_score FROM olist_order_reviews r WHERE r.order_id = oo.order_id) = 5;'''
df_sql = pd.read_sql_query(query,con=engine)
df_sql.head()
```


 Query 2: Customer Purchase Trend Year-on-Year

Write and execute a SQL query to analyze the customer purchase trend year-on-year.

```python
query='''SELECT strftime('%Y', oo.order_purchase_timestamp) AS year,
       COUNT(DISTINCT oo.customer_id) AS customer_count
FROM olist_orders oo
GROUP BY year
ORDER BY year;'''
df_sql = pd.read_sql_query(query,con=engine)
df_sql.head()
```

5. Query 3: Average Order Values of Customers

Write and execute a SQL query to calculate the average order values of customers.

```python
query ='''SELECT DISTINCT c.customer_unique_id, AVG(oi.price + oi.freight_value) AS avg_order_value
FROM olist_customers c
JOIN olist_orders o ON c.customer_id = o.customer_id
JOIN olist_order_items oi ON o.order_id = oi.order_id  GROUP BY c.customer_unique_id;'''
df_sql = pd.read_sql_query(query,con=engine)
df_sql.head()
```

6. Query 4: Top 5 Cities with Highest Revenue from 2016 to 2018

Write and execute a SQL query to find the top 5 cities with the highest revenue from 2016 to 2018.

```python
query='''SELECT c.customer_city, SUM(oi.price + oi.freight_value) AS total_revenue
FROM olist_customers c
JOIN olist_orders o ON c.customer_id = o.customer_id
JOIN olist_order_items oi ON o.order_id = oi.order_id
WHERE (SELECT strftime('%Y', o.order_purchase_timestamp) FROM olist_orders)
BETWEEN '2016' AND '2018' AND oi.order_item_id IS NOT NULL
GROUP BY c.customer_city
ORDER BY total_revenue DESC
LIMIT 5;'''
df_sql = pd.read_sql_query(query,con=engine)
df_sql
```

7. Query 5: State Wise Revenue Table Between 2016 to 2018

Write and execute a SQL query to create a state-wise revenue table between 2016 to 2018.

```python
query='''SELECT c.customer_state, SUM(oi.price + oi.freight_value) AS total_revenue
FROM olist_customers c
JOIN olist_orders o ON c.customer_id = o.customer_id
JOIN olist_order_items oi ON o.order_id = oi.order_id
WHERE (SELECT strftime('%Y', o.order_purchase_timestamp) FROM olist_orders)
BETWEEN '2016' AND '2018' AND oi.order_item_id IS NOT NULL
GROUP BY c.customer_state;'''
df_sql =pd.read_sql(query,con=engine)
df_sql.head()
```

. Query 6: Top Successful Sellers in Terms of Goods Sold, Revenue, and Customer Count

Write and execute a SQL query to identify the top successful sellers in terms of the number of goods sold, total revenue, customer count, and sellers with the highest 5-star ratings.

```python
query='''SELECT s.seller_id, COUNT(DISTINCT oi.order_id) AS goods_sold,
SUM(oi.price + oi.freight_value) AS total_revenue,
COUNT(DISTINCT o.customer_id) AS customer_count FROM olist_sellers s
 JOIN olist_order_items oi ON s.seller_id = oi.seller_id
 JOIN olist_orders o ON oi.order_id =o.order_id
WHERE(SELECT r.review_score FROM olist_order_reviews r WHERE r.order_id = oi.order_id) = 5
GROUP BY s.seller_id ORDER BY goods_sold DESC, total_revenue DESC, customer_count DESC LIMIT 5;'''
df_sql =pd.read_sql(query,con=engine)
df_sql.head()
```

 Query 7: Delivery Success Rate Across States

Write and execute a SQL query to calculate the delivery success rate across different states.


```python
query='''WITH deliveries AS (SELECT c.customer_state AS state,
COUNT(DISTINCT CASE WHEN o.order_status = 'delivered' THEN o.order_id END) AS delivered_orders,
COUNT(DISTINCT o.order_id) AS total_orders
FROM olist_customers c
JOIN olist_orders o ON c.customer_id = o.customer_id
JOIN olist_order_items oi ON o.order_id = oi.order_id GROUP BY c.customer_state)
SELECT state, delivered_orders, total_orders ,
(delivered_orders*100)/total_orders AS delevery_rate FROM deliveries;'''
df_sql =pd.read_sql(query,con=engine)
df_sql.head()
```

10. Query 8: Preferred Form of Payment for Different Categories

Write and execute a SQL query to find the preferred form of payment for different product categories.


```python
query='''SELECT p.product_category_name, op.payment_type, COUNT(op.payment_type) AS payment_count
FROM olist_products_dataset p
JOIN olist_order_items oi ON p.product_id = oi.product_id
JOIN olist_order_payments op ON oi.order_id = op.order_id
GROUP BY p.product_category_name, op.payment_type
ORDER BY p.product_category_name, payment_count DESC;'''
df_sql =pd.read_sql(query,con=engine)
df_sql.head()
```

11. Query 9: Distance Between Cities

Write and execute a SQL query to calculate the distance between cities.


```python
query='''
WITH CityLocations AS (
  SELECT
    geolocation_city,
    AVG(geolocation_lat) AS avg_lat,
    AVG(geolocation_lng) AS avg_lng
  FROM
    olist_geolocation
  GROUP BY
    geolocation_city
)
SELECT
  cl1.geolocation_city AS city1,
  cl2.geolocation_city AS city2,
  (6371 * ACOS(
    COS(cl1.avg_lat * 3.141592653589793 / 180.0) * COS(cl2.avg_lat * 3.141592653589793 / 180.0) *
    COS(cl2.avg_lng * 3.141592653589793 / 180.0 - cl1.avg_lng * 3.141592653589793 / 180.0) +
    SIN(cl1.avg_lat * 3.141592653589793 / 180.0) * SIN(cl2.avg_lat * 3.141592653589793 / 180.0)
  )) AS distance_km
FROM
  CityLocations AS cl1
JOIN
  CityLocations AS cl2 ON cl1.geolocation_city < cl2.geolocation_city
WHERE
  cl1.geolocation_city IS NOT NULL AND cl2.geolocation_city IS NOT NULL
GROUP BY
  cl1.geolocation_city, cl2.geolocation_city
ORDER BY
  distance_km DESC
LIMIT 10;'''
df_sql =pd.read_sql(query,con=engine)
df_sql.head()
```
