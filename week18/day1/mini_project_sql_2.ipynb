```python
import pandas as pd
import sqlite3
```

```python
df_flix=pd.read_csv('netflix_titles.csv')
df_flix.head()
```

```python
import sqlalchemy
engine = sqlalchemy.create_engine('sqlite:///netflix.db')
df_flix.to_sql('netflix', engine, index=False)
```

Query 13: Case Statements

Write and execute a SQL query to use CASE statements for conditional logic.

```python
query = '''SELECT title, release_year,
 CASE
 WHEN release_year > 2015 THEN 'current_movie' ELSE 'old_movie' END AS movie_status FROM netflix;'''
df_sql=pd.read_sql(query, engine)
df_sql.head()
```

Query 14: Sub Queries

Write and execute a SQL query to use a subquery for selection.

```python
query ='''SELECT title, release_year , rating FROM netflix
 WHERE rating IN (SELECT rating FROM netflix WHERE rating = 'PG-13')'''
df_sql=pd.read_sql(query, engine)
df_sql.head()
```

 Query 15: Coalesce

Write and execute a SQL query to use COALESCE to handle null values.

```python
query= '''SELECT title, COALESCE(country, 'Unknown') AS country,
 COALESCE(director, 'unknown') AS director
 FROM netflix'''
df_sql=pd.read_sql(query, engine)
df_sql.head()
```

Query 16: Convert

Write and execute a SQL query to convert data types.

```python
query ='''SELECT title, CAST(release_year AS TEXT) AS release_year FROM netflix'''
df_sql=pd.read_sql(query, engine)
df_sql.head()
```

 Query 17: Lag/Lead

Write and execute a SQL query to use LAG/LEAD for window functions.

```python
query ='''SELECT title, release_year,
 CAST(LAG(release_year) OVER (ORDER BY release_year) AS INTEGER) AS previous_year,
 CAST(LEAD(release_year) OVER (ORDER BY release_year) AS INTEGER) AS next_year
 FROM netflix'''
df_sql=pd.read_sql(query, engine)
df_sql.head()
```

Query 18: Row Number

Write and execute a SQL query to use ROW_NUMBER for window functions.

```python
query ='''SELECT title, release_year,
 ROW_NUMBER() OVER (ORDER BY release_year) AS row_num
 FROM netflix'''
df_sql=pd.read_sql(query, engine)
df_sql.head()
```

 Query 19: Dense Rank

Write and execute a SQL query to use DENSE_RANK for window functions.

```python
query ='''SELECT title, release_year,
 DENSE_RANK() OVER (ORDER BY release_year) AS dense_rank
 FROM netflix'''
df_sql=pd.read_sql(query, engine)
df_sql.head()
```

Query 20: With

Write and execute a SQL query to use WITH clause for common table expressions (CTE).


```python
query='''WITH current_movies AS (SELECT title, director, release_year
FROM netflix WHERE release_year > 2015)
SELECT * FROM current_movies'''
df_sql=pd.read_sql(query, engine)
df_sql.head()
```
