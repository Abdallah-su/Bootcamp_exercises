```python
from urllib.request import urlopen, Request
from bs4 import BeautifulSoup
import requests
import shutil
from selenium.common.exceptions import WebDriverException
import pandas as pd
import time
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options 
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC


```

```python
# install required packages (run in this notebook environment)
%pip install -q selenium webdriver-manager

options = Options()
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--ignore-certificate-errors") # Ignores SSL issues
options.add_argument("--proxy-server='direct://'")
options.add_argument("--proxy-bypass-list=*")

# if Chrome is not installed in default location, try to find the binary
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install(),options =options))
```

```python
url ='https://www.inmotionhosting.com/dedicated-servers'
driver.get(url)
```

```python
print(driver.title)
soup =BeautifulSoup(driver.page_source, 'html.parser')
```

```python
details = [ ]
print(len(soup.find_all('div', class_ ="imh-product-rostrum product pill-switcher-style mobile-slider rostrum_scrolling_switcher")))
for detail  in soup.find_all('div', class_ ='imh-rostrum-card'):
  detail_dict = {'name': detail.find('h3', class_ ="imh-rostrum-card-title").get_text(strip=True),
                 'features': ','.join(f.get_text(strip=True) for f in soup.find_all('span', "imh-rostrum-detail tooltips-rostrum")),
                 'price': detail.find('span', class_="rostrum-price").get_text(strip=True)}
  details.append(detail_dict)
df_host_details = pd.DataFrame(details)
df_host_details.to_csv('host_details.csv', index=False)
df_host_details.head()


```
