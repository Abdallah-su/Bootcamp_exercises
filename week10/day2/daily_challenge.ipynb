1. Data Preparation:

Hint 1: Use np.random.uniform(low, high, size) to generate the temperature data.
Hint 2: Create a DataFrame using pd.DataFrame(data, index, columns) with appropriate index and columns.

Use NumPy to generate a synthetic dataset representing average monthly temperatures (in degrees Celsius) for 12 months across 10 different cities. The temperatures should range from -5 to 35 degrees.

Convert this NumPy array into a Pandas DataFrame, adding city names as index and months as columns.


```python
import numpy as np
import pandas as pd
np.random.seed(1)
temp_data = np.random.uniform(-5, 35, (10, 12)).round(2)
months =['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
cities =['Accra', 'Wenchi','Kumasi','Tamale','Sunyani','Goaso','Tepa','Obuasi','Tema','Berekum']
temp_df = pd.DataFrame(data =temp_data, index= cities, columns = months)
temp_df.head(10)

```

```python
temp_df.columns
```

. Data Analysis:

Hint 1: Calculate the annual average temperature using DataFrame.mean(axis).
Hint 2: Find the city with the highest and lowest average temperature using idxmax() and idxmin() methods.

Calculate the annual average temperature for each city.

Identify the city with the highest and lowest average temperature for the year.

```python

print(f'the anuall averages per city; {temp_df.mean(axis =1).round(2)}')
```

```python
city_temp = temp_df.mean(axis =1)
city_high =city_temp.idxmax()
print(f'the hottest city is {city_high}')
city_low =city_temp.idxmin()
print(f'the coldest city is {city_low}' )
```

3. Data Visualization:

```python
import seaborn as sns
import matplotlib.pyplot as plt
new_temp= temp_df.reset_index().melt(id_vars ='index',var_name='months', value_name ='temperature')
new_temp.columns =['city','months','temperature']

sns.lineplot(data =new_temp, x= 'months', y='temperature' , hue ='city')
plt.title('Monthly Temperature Trends Across Cities')
plt.ylabel('Temperature')
plt.legend(loc='lower left')
plt.show()
```

The hottest city from the analysis is Berekum with an average temperature of 18.2 degrees celcius and coldest is Sunyani with 8.82. Most of the cities experience cold in the month of March
