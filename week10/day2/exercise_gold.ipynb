
Exercise 1 : Minimum and Maximum of Random Array
Instructions
Create a 5x5 array with random values and find the minimum and maximum values.

```python
import numpy as np
np.random.seed(1)
array = np.random.randint(1, 100, (5,5))
print(f'mininum value : {np.min(array)}, maximum value : {np.max(array)}')
```

Exercise 2 : Matrix Normalization
Instructions
Normalize a 3x3 random matrix (subtract the mean and divide by the standard deviation of the matrix).

```python
np.random.seed(1)
array =np.random.randint(1, 50, (3,3))
std = np.std(array)
avg = np.mean(array)
norm_array =((array -avg)/std).round(2)
print(f'array : {array :}')
print('')
print(f'''normalised array :
{norm_array}''')
```
